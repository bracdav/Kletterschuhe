```python
baseurl = 'https://www.bergzeit.ch/'
tag = 'products-list-page-preloader__links'
```

```python
scraper = ScrapExecutor()
links = []
for page in range(1, 8):
    links.extend(scraper.execute(f'https://www.bergzeit.ch/schuhe/kletterschuhe/?p={page}&sortBy=ct',
                                 baseurl, tag))
```

```python
a= scraper.execute(links, baseurl)
```

```python
import pandas as pd

df = pd.DataFrame(a)

df.to_csv('ScrapedData/kletterschuhe_raw.csv')
```
