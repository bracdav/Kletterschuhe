```python
baseurl = 'https://www.bergzeit.ch/'
tag = 'products-list-page-preloader__links'
```

```python
scraper = ScrapExecutor()
links = []
for page in range(1, 8):
    links.extend(scraper.execute(f'https://www.bergzeit.ch/schuhe/kletterschuhe/?p={page}&sortBy=ct',
                                 baseurl, tag))
```

```python
a= scraper.execute(links, baseurl)
```

```python
import pandas as pd

df = pd.DataFrame(a)
df.head()
```

```python
import numpy as np

def flatten_lists(df):
    res = []
    for sizes in df['availableSizes']:
        for size in sizes:
            if size not in res:
                res.append(size)

    return np.unique(res)

def convert_size(flat_list):
    res = {}
    for size in flat_list:
        if 'US' in size:
            res[size] = np.round(float(size.replace('US ', '').replace(',','.').replace('+','').strip()), 1) + 32
        elif 'UK' in size:
            res[size] = np.round(float(size.replace('UK ', '').replace(',','.').strip()), 1) + 33
        elif '-' in size:
            res[size] = np.round(float(size.split('-')[0]), 1)
        elif 'EUI' in size:
            res[size] = np.round(float(size.replace('EUI ', '').replace(',', '')), 1)
        else:
            res[size] = np.round(float(size.replace(',', '')), 1)
    return res

a = flatten_lists(df)

mapper = convert_size(a)

df['availableSizes'].map(mapper)
```

```python
# 1. Find all columns where at least one entry is a list
list_cols = [col for col in df.columns if df[col].apply(lambda x: isinstance(x, list)).any()]

# 2. Loop through those columns and expand them
for col in list_cols:
    # .apply(pd.Series) splits each list element into its own column
    expanded = df[col].apply(pd.Series)
    
    # Rename the new columns to avoid naming collisions (e.g., features_0, features_1)
    expanded = expanded.rename(columns=lambda x: f"{col}_{x}")
    
    # Drop the original list column and merge the new columns
    df = pd.concat([df.drop(columns=[col]), expanded], axis=1)
```
